```python
import  micropip

# Install required packages using micropip
await micropip.install("numpy")
await micropip.install("matplotlib")
await micropip.install("scipy")

# Import installed packages
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# Create figure with larger size for better visibility
plt.figure(figsize=(15, 9))

# Generate x values for smooth curve visualization
x = np.linspace(-4, 4, 1000)
y = stats.norm.pdf(x, loc=0, scale=1)

# Plot main distribution curve with increased line width
plt.plot(x, y, 'b-', linewidth=3, label='Skill Distribution')

# Define region boundaries and colors
sigma_scholars_left = 2
mids_right = 2
mids_left = -2
skibs_right = -2

# Add shaded regions with improved contrast
plt.fill_between(x[x >= sigma_scholars_left], 
                y[x >= sigma_scholars_left],
                alpha=0.4, color='#FF4444',  # Bright red
                label='SIGMA SCHOLARS')
plt.fill_between(x[(x <= mids_right) & (x >= mids_left)],
                y[(x <= mids_right) & (x >= mids_left)],
                alpha=0.4, color='#FFFF44',  # Bright yellow
                label='MIDS')
plt.fill_between(x[x <= skibs_right],
                y[x <= skibs_right],
                alpha=0.4, color='#654321',  # Dark brown
                label='SKIBS')

# Add vertical boundary lines with increased thickness
plt.axvline(x=sigma_scholars_left, color='#FF4444', linestyle='--', linewidth=2.5)
plt.axvline(x=mids_right, color='#888888', linestyle='--', linewidth=2.5)
plt.axvline(x=mids_left, color='#888888', linestyle='--', linewidth=2.5)
plt.axvline(x=skibs_right, color='#654321', linestyle='--', linewidth=2.5)

# Add large, clear region labels with improved positioning
plt.text(2.7, 0.08, 'SIGMA SCHOLARS\n(>2σ)', fontsize=24, 
         color='#FF4444', fontweight='bold',
         horizontalalignment='center')
plt.text(0, 0.40, 'MIDS\n(-2σ to +2σ)', fontsize=28,
         color='#000000', fontweight='bold',
         horizontalalignment='center')
plt.text(-2.7, 0.08, 'SKIBS\n(<-2σ)', fontsize=24,
         color='#654321', fontweight='bold',
         horizontalalignment='center')

# Enhance plot appearance for accessibility
plt.title('Distribution of STEM Skills', fontsize=32, pad=30)
plt.xlabel('Skill Level (Standard Deviations from Mean)', fontsize=20)
plt.ylabel('Frequency', fontsize=20)
plt.grid(True, alpha=0.3, linestyle='--')
plt.legend(loc='upper left', fontsize=18)

# Adjust layout to prevent label clipping
plt.tight_layout()

# Display the plot
plt.show()
```
